# GCD-EDU — comparación inicial de políticas

Este notebook ejecuta el simulador sintético de referencia y compara cuatro políticas. **No es evidencia de eficacia educativa**: los resultados dependen de las reglas del entorno simulado.

Antes de ejecutar: instala el paquete con `python -m pip install -e '.[notebooks]'` desde la raíz.

```python
from pathlib import Path
import sys

# Permite ejecutar el notebook desde el repositorio sin instalar el paquete.
ROOT = Path.cwd()
if (ROOT / 'src' / 'gcd_edu').exists():
    sys.path.insert(0, str(ROOT / 'src'))
elif (ROOT.parent / 'src' / 'gcd_edu').exists():
    sys.path.insert(0, str(ROOT.parent / 'src'))

import pandas as pd
import matplotlib.pyplot as plt
from gcd_edu.simulator import run_comparison, summarize
```

```python
SEED = 20261015
N_REPLICATES = 200
STEPS = 30
DELAY_STEPS = 8

results = run_comparison(
    n_replicates=N_REPLICATES,
    steps=STEPS,
    delay_steps=DELAY_STEPS,
    seed=SEED,
)
summary = summarize(results)
summary.round(3)
```

```python
metrics = ['immediate_mastery', 'delayed_retention', 'transfer_score']
display_cols = ['policy'] + [f'{m}_mean' for m in metrics] + [f'{m}_std' for m in metrics]
summary[display_cols].sort_values('delayed_retention_mean', ascending=False).round(3)
```

```python
metric = 'delayed_retention'
ax = (results.groupby('policy')[metric]
      .agg(['mean', 'std'])
      .sort_values('mean')
      .plot(kind='bar', y='mean', yerr='std', legend=False,
            title='Retención diferida sintética por política',
            ylabel='Puntuación simulada (0–1)', xlabel='Política'))
ax.set_ylim(0, 1)
plt.tight_layout()
plt.show()
```

## Interpretación y límites

- La puntuación es generada por reglas sintéticas, no por estudiantes reales.
- Las diferencias describen este entorno artificial; no prueban superioridad general.
- La política adaptativa y los controles comparten una estructura de acciones, pero no necesariamente un presupuesto idéntico de utilidad pedagógica.
- El siguiente paso metodológico es desacoplar los perfiles latentes de los flujos aleatorios de cada política y añadir escenarios en los que la adaptación no tenga ventaja.
